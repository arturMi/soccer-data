```python
import pandas as pd
```

```python
data_df = pd.read_csv('data/data.csv', sep=',')
print(data_df.columns.values.tolist())
print(data_df.shape)
```

```python
team_df = pd.read_csv('data/teams_data.csv', sep=',')
print(team_df.columns.values.tolist())
print(team_df.shape)
```

```python
home_team_df = pd.merge(data_df, team_df,  how='inner', left_on=['home_team_id','date'], right_on = ['team_id','date'])
print(home_team_df.columns.values.tolist())
print(home_team_df.shape)
```

```python
away_team_df = pd.merge(data_df, team_df,  how='inner', left_on=['away_team_id','date'], right_on = ['team_id','date'])
print(away_team_df.columns.values.tolist())
print(away_team_df.shape)
```

```python
df = pd.merge(home_team_df, away_team_df,  how='inner', left_on=['home_team_id','away_team_id','date'], right_on = ['home_team_id','away_team_id','date'])
print(df.columns.values.tolist())
print(df.shape)
```
